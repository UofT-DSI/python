Part - 1 Testing the Anagram without considering the Case senstivity.

```python
def anagram_checker(word_a, word_b):
    """Return True if word_a and word_b are anagrams, ignoring case senstivity."""
    # Treat uppercase and lowercase as the same letter.
    lowercase_a = word_a.lower()
    lowercase_b = word_b.lower()

    # Same letters in any order means the sorted character lists match.
    return sorted(lowercase_a) == sorted(lowercase_b)
```

Part 1 - Testing

```python
print(anagram_checker("Silent", "listen")) 
```

```python
print(anagram_checker("Silent", "Night")) 
```

```python
print(anagram_checker("night", "Thing"))  
```

Part 2 : Expanding the functionality of the ANgram Checker

```python
def anagram_checker(word_a, word_b, is_case_sensitive):
    """Return True if the two words are anagrams.

    is_case_sensitive:
        False -> uppercase and lowercase count as the same letter.
        True  -> letters must match in case as well as identity.
        
    """
    if is_case_sensitive:
        lowercase_a = word_a
        lowercase_b = word_b
    else:
        lowercase_a = word_a.lower()
        lowercase_b = word_b.lower()

    # Same multiset of characters means the sorted lists are equal.
    return sorted(lowercase_a) == sorted(lowercase_b)
```

Part 2 : Testing the output 

```python
print(anagram_checker("Silent", "listen", False))
```

```python
print(anagram_checker("Silent", "listen", True))
```

```python
print(anagram_checker("Silent", "Listen", True))
```
