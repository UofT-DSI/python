## Filtering and selecting data

You can filter data by passing a series of true / false to use as indices. 
You can also use .loc to select by index explicitly

````python
df.loc[~df["bike_model"].isna()].head() # select only rows where bike_model is not NaN
````

## Grouping

````python
df.groupby("occurrence_dayofweek")['cost_of_bike'].mean()
````

````python
df.columns
````

````python
mean_costs_by_date = df.groupby(["occurrence_month", 
                                 "occurrence_dayofweek"])[['cost_of_bike', 'bike_speed']].mean().reset_index()
````

````python
mean_costs_by_date.sort_values(by = 'cost_of_bike')
````

groupby is very powerful! We have only scratched the surface. You're doing SQL next. Know that all the operations you can do in SQL - grouping, aggregating, joining data, etc. - you can do this with pandas as well.

## Saving to file

````python
mean_costs_by_date.to_csv("bike_avg_cost_by_date.csv", index = None) 
````
