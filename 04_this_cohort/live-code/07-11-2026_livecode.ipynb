```python
def calc_total_bill(price, tax_rate = 0.13, tip_rate = 0.20):
    """
    Calculate the total bill including tax and tip.
    Parameters: 
    price (float): The price of a meal.
    tax_rate (float): The rate of tax as a decimal.
    tip_rate (float): The rate of tip as a decimal.
    Returns:
    The total bill including tax and tip (float).
    """
    tax = price * tax_rate
    tip = price * tip_rate
    return price + tax + tip
```

```python
calc_total_bill(100)
```

```python
calc_total_bill(price = 100, tip_rate = 0.22)
```

```python
calc_total_bill(100, 0.22)
```

```python
def moles (mass, molar_mass):
    """
    Calculate the moles of a compound or element from its mass and molar_mass
    Parameters:
    mass (float): Mass in grams (g)
    molar_mass (float): Molar mass in grams per mol (g/mol)
    Returns:
    float: Moles of a given compound
    """
    return round((mass/molar_mass), 2)
```

```python
def oil_change(current_km):
    """
    This function takes the exact kilometers from your last oil change and returns the next oil‑change due.    
    """
    return current_km + 5000
```

```python
def add_1(integer):
    """
    Adds 1 to the given integer to calculate what is your number in queue by knowing how many people are in front of you.
    """
    return integer + 1
```
