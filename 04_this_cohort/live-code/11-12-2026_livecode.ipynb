```python
def temp_input():
    temp = int(input("What is the temperature in degrees Celsius?"))

    if temp < 0:
        print("It is freezing cold!")
    elif temp <= 15:
        print("It is cold!")
    elif temp <= 30:
        print("It is hot!")
    elif temp <= 45:
        print("It is really hot!")
    else:
        print("It is extremely hot!")
```
