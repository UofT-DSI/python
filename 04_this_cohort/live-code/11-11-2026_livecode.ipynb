```python
def calc_total_bill(price, tax_rate = 0.13, tip_rate = 0.20):
    tax = price * tax_rate
    tip = price * tip_rate
    return price + tax + tip
```

```python
calc_total_bill(100)
```

```python
calc_total_bill(100, tip_rate = 0.22)
```

```python
calc_total_bill(100, 0.22)
```

```python
def c_to_f(degrees_c):
    """
    Converts a temperature from Celsius to Fahrenheit.

    Parameters:
    degrees_c (float): Temperature in degrees Celsius

    Return:
    float: Temperature in degrees Fahrenheit
    """
    return (degrees_c *(9/5)) + 32
```

```python
def tot_pts(goals, assists):
    """
    Calculate the total points
    Paramters:
    goals (int): number of goals
    assists (int): number of assists
    """
    return goals + assists
```

```python
tot_pts(5,2)
```

```python
def BMI_calculator(height, weight):
    """
    This function calculates your BMI based on height and weight
    Paramters:
    height(float): in meters
    weight (float): in kg
    """
    return weight/height**2
```

```python
BMI_calculator(1.78, 85)
```
