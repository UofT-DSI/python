```python
#set one

rainbow = {'red','orange','yellow','green','blue','indigo','violet'}

#set two

olympic_flag = {'red','green','yellow','blue','black'}
```

```python
#union = combines the two sets, to give us all unique values from both

rainbow.union(olympic_flag)
```

```python
#intersection = finds common values from both sets

rainbow.intersection(olympic_flag)
```

```python
#symmetric difference finds values are in set one OR set two but not both

rainbow.symmetric_difference(olympic_flag)
```

```python
#difference find values unique to the first set only

rainbow.difference(olympic_flag)
```

```python
olympic_flag.difference(rainbow)
```
