```python
def temp():
    current_temp = float(input("What is the current temperature"))
    if current_temp <= 0:
        print("It is freezing cold!")
    elif current_temp <= 15:
        print("It is cold!")
    elif current_temp <= 30:
        print("It is hot!")
    elif current_temp <= 45:
        print("It is really hot!")
    else:
        print("It is extremely hot!")
```

```python
temp()
```
