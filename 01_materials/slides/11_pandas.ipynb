We can rename any columns in our DataFrame that may not be so intuitive. 
`rename()`  takes a dictionary with the current column name as the key and the new name as the value.
This doesn't modify the DataFrame in-place, so if we want to keep it, we must reassign the result to `df` (or another variable).
Running this same code again doesn't give an error, it just doesn't do anything.

````python
df = df.rename(columns = {"Cost_of_Bike": "Bike_Cost" })
````

````python
# the Occurence_Date column is supposed to have dates and times, so lets make it a 
# "datetime" data type to make it easier to work with

df["Occurrence_Date"]
````

In pandas, most operations return a new

````python
df.drop(columns = '_id') # returns the dataframe, can reassign it if desired.
````

````python
# can create new columns by just calling the new column name
df["current_year"] = 2026
df["Occurrence_Month"] = df["Occurrence_Month"].str[:3] # change to just the first 3 characters
````

## Filtering and selecting data

You can filter data by passing a series of true / false to use as indices. 
You can also use .loc to select by index explicitly

````python
# 
df["Bike_Model"].isna() # returns a series of true / false
df.loc[df["Bike_Model"].isna()] # use that series as an index
````

````python
df.loc[0:10] # return the row with index 0:10 
#.loc uses the actual index, #.iloc uses the position index
# careful: sometimes the index values don't match the row position
````

````python
# We can use the .loc[]and .isin() to filter rows in a DataFrame by providing a
# list of values for a column
models_to_keep = ["HYBRID", "SIRRUS", "ESCAPE"]

df.loc[df["Bike_Model"].isin(models_to_keep)]
````

## Grouping

````python
df.columns
````

````python
# groupby: create groups based on a column or columns, then apply summarizations to them
df.groupby("Occurrence_DayOfWeek")['Bike_Cost'].mean()
````

````python
# can group by multiple columns
# if you want the result to be its own dataframe, can use reset_index()

mean_costs_by_date = df.groupby(["Occurrence_Month", 
                                 "Occurrence_DayOfWeek"])['Bike_Cost'].mean().reset_index()

# can sort columns using sort_values
mean_costs_by_date.sort_values(by = 'Bike_Cost')
````

````python
# you can also use groupby() to create an object to iterate over

for colour, group in df.groupby("Bike_Colour"):
    # group is just a smaller dataframe with all like Bike_Colour
    print(colour, group['Bike_Cost'].mean()) 
````

groupby is very powerful! We have only scratched the surface. You're doing SQL next. Know that all the operations you can do in SQL - grouping, aggregating, joining data, etc. - you can do this with pandas as well.

````python
# saving to file

# I like to pass index = None, otherwise you get a new index column in the saved file
mean_costs_by_date.to_csv("bike_avg_cost_by_date.csv", index = None) 
````
